```python
# Word Count - MapReduce simulation (Colab)
from collections import defaultdict
import re


text = """
This is a sample text. This text is for Word Count example.
MapReduce style: split by words, map to (word,1), then group & reduce.
"""

# Mapper
def mapper_line(line):
    words = re.findall(r"\b\w+\b", line.lower())
    return [(w, 1) for w in words]

# Shuffle/Group
def group(pairs):
    grouped = defaultdict(list)
    for k, v in pairs:
        grouped[k].append(v)
    return grouped

# Reducer
def reducer(grouped):
    reduced = {}
    for k, vals in grouped.items():
        reduced[k] = sum(vals)
    return reduced


pairs = []
for line in text.strip().splitlines():
    pairs.extend(mapper_line(line))

grouped = group(pairs)
result = reducer(grouped)

for word, count in sorted(result.items(), key=lambda x: -x[1])[:20]:
    print(f"{word}: {count}")
```

```python
import csv
from collections import defaultdict

file_path = "/content/drive/MyDrive/sales.csv"

# Mapper:
def mapper_sales(row):

    revenue = float(row['Sales'])
    region = row['Market']
    return [(region, revenue)]

# Grouping
def group(pairs):
    grouped = defaultdict(list)
    for k, v in pairs:
        grouped[k].append(v)
    return grouped

# Reducer
def reducer_sum(grouped):
    return {region: sum(values) for region, values in grouped.items()}

pairs = []

with open(file_path, "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        pairs.extend(mapper_sales(row))

# Group
grouped = group(pairs)

# Reduce
result = reducer_sum(grouped)

# Display results
print("Total Revenue per Region:")
for region, total in result.items():
    print(f"{region} → {total:.2f}")
```

```python

from collections import defaultdict
import os


logfile_path = "/content/weblogs.txt"


def mapper(line):
    line = line.strip()
    if not line:
        return []

    parts = line.split(",")

    if len(parts) < 6:

        return []

    status = parts[5].strip()
    if status == "":
        return []
    return [(status, 1)]

# Shuffle/Group
def group(pairs):
    grouped = defaultdict(list)
    for k,v in pairs:
        grouped[k].append(v)
    return grouped

# Reducer:
def reducer_count(grouped):
    counts = {}
    for k, vals in grouped.items():
        counts[k] = sum(vals)
    return counts

pairs = []
if os.path.exists(logfile_path):
    with open(logfile_path, "r", encoding="utf-8") as f:
        for line in f:
            pairs.extend(mapper(line))
else:

    sample_logs = [
        "2025-10-10,12:01:32,192.168.1.2,GET,/index.html,200,1024",
        "2025-10-10,12:01:33,192.168.1.3,GET,/products.html,200,850",
        "2025-10-10,12:01:35,192.168.1.4,GET,/contact.html,404,512",
        "2025-10-10,12:01:36,192.168.1.5,POST,/login,500,30",
        "2025-10-10,12:01:37,192.168.1.6,GET,/index.html,200,1024",
        "2025-10-10,12:01:38,192.168.1.7,GET,/forbidden.html,403,0",

    ]
    for line in sample_logs:
        pairs.extend(mapper(line))

grouped = group(pairs)
counts = reducer_count(grouped)


for status in sorted(counts.keys()):
    print(f"HTTP {status}: {counts[status]} requests")
```

```python
def mapper_url(line):
    parts = line.strip().split(",")
    if len(parts) < 5:
        return []
    url = parts[4].strip()
    return [(url, 1)] if url else []
```

```python
def mapper_status_size(line):
    parts = line.strip().split(",")
    if len(parts) < 7:
        return []
    status = parts[5].strip()
    try:
        size = int(parts[6].strip())
    except:
        size = 0
    return [(status, size)]
```
